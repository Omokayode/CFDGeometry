Draw a rectangle on the map, click **Use this extent**, then run the download cell.

pip install -e ".[notebook,download]"

````python
from cfd_geometry.notebook import select_extent

selector = select_extent(place="Milwaukee, Wisconsin, USA")
selector
````

````python
from pathlib import Path

from cfd_geometry.download import DownloadConfig, download_domain

if selector.bbox is None:
    raise RuntimeError(
        "Draw a rectangle on the map and click 'Use this extent' first."
    )

cfg = DownloadConfig(
    output_dir=Path("data/input"),
    bbox=selector.bbox,
    layers=("buildings", "trees"),
    download_dem=False,
)
result = download_domain(cfg)
result.bbox
````
